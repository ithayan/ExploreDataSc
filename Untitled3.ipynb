````python
# FULL COLAB PIPELINE: Indian-Medicine + pseudo-3D + FAISS RAG (no RDKit)

!pip install -q pandas torch torchvision sentence-transformers requests faiss-cpu

import os, time
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from sentence_transformers import SentenceTransformer
import faiss

# =========================
# 0. LOAD CSV FROM /content
# =========================
CSV_PATH = "/content/indian_medicine_data.csv"  # upload file via left panel
df = pd.read_csv(CSV_PATH)
print("Loaded rows:", len(df))

name_col = "name"
comp1_col = "short_composition1"
comp2_col = "short_composition2"

df_simple = df[[name_col, comp1_col, comp2_col]].fillna("")
df_simple["composition"] = df_simple[comp1_col] + "; " + df_simple[comp2_col]
df_simple = df_simple[[name_col, "composition"]].head(1000).reset_index(drop=True)
print(df_simple.head())

# =========================
# 1. TEXT EMBEDDINGS
# =========================

text_model = SentenceTransformer("all-MiniLM-L6-v2")

corpus = [
    f"Product: {row[name_col]}. Composition: {row['composition']}."
    for _, row in df_simple.iterrows()
]
text_embeddings = text_model.encode(corpus, convert_to_tensor=True)
text_emb = text_embeddings.detach().cpu()
text_emb = torch.tensor(text_emb, dtype=torch.float32)   # (N, d_text)
print("Text embedding shape:", text_emb.shape)

# =========================
# 2. INGREDIENT EXTRACTION
# =========================

def extract_ingredients(series):
    names = []
    for line in series:
        if not isinstance(line, str):
            continue
        for part in line.split("+"):
            for token in part.split("/"):
                n = token.strip()
                if n and not any(c.isdigit() for c in n):
                    names.append(n.lower())
    return sorted(set(names))

all_comp = df[[comp1_col, comp2_col]].fillna("").agg("; ".join, axis=1)
unique_ingredients = extract_ingredients(all_comp)
print("Unique ingredient count:", len(unique_ingredients))
print("Examples:", unique_ingredients[:20])

# =========================
# 3. PSEUDO-3D EMBEDDINGS (CHAR CNN ON INGREDIENT NAME)
# =========================

alphabet = "abcdefghijklmnopqrstuvwxyz0123456789-(), "
char_to_idx = {c: i + 1 for i, c in enumerate(alphabet)}  # 0=pad
vocab_size = len(char_to_idx) + 1
max_len = 40

def encode_string(s):
    s = s.lower()[:max_len]
    arr = np.zeros(max_len, dtype=np.int64)
    for i, ch in enumerate(s):
        arr[i] = char_to_idx.get(ch, 0)
    return arr

class Simple1DCNN(nn.Module):
    def __init__(self, vocab_size, emb_dim=32, out_dim=128):
        super().__init__()
        self.embed = nn.Embedding(vocab_size, emb_dim, padding_idx=0)
        self.conv1 = nn.Conv1d(emb_dim, 64, kernel_size=3, padding=1)
        self.conv2 = nn.Conv1d(64, 64, kernel_size=3, padding=1)
        self.fc = nn.Linear(64, out_dim)

    def forward(self, x):  # (B,L)
        x = self.embed(x).transpose(1, 2)        # (B,emb,L)
        x = F.relu(self.conv1(x))
        x = F.relu(self.conv2(x))
        x = F.adaptive_max_pool1d(x, 1).squeeze(-1)  # (B,64)
        x = self.fc(x)
        x = F.normalize(x, dim=1)
        return x

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model_1d = Simple1DCNN(vocab_size=vocab_size, emb_dim=32, out_dim=128).to(device)

ing_to_emb = {}
with torch.no_grad():
    for name in unique_ingredients:
        token_ids = torch.tensor(encode_string(name))[None, :].to(device)
        emb = model_1d(token_ids).cpu().squeeze(0)  # (128,)
        ing_to_emb[name] = emb

print("Pseudo-3D embeddings for ingredients:", len(ing_to_emb))

# =========================
# 4. AGGREGATE PER PRODUCT
# =========================

def product_3d_embedding(composition_text):
    if not isinstance(composition_text, str):
        return None
    parts = []
    for part in composition_text.split("+"):
        for token in part.split("/"):
            n = token.strip().lower()
            if n and not any(c.isdigit() for c in n) and n in ing_to_emb:
                parts.append(ing_to_emb[n])
    if not parts:
        return None
    return torch.stack(parts, dim=0).mean(dim=0)

mol_emb_list = []
for comp in df_simple["composition"]:
    emb = product_3d_embedding(comp)
    if emb is None:
        emb = torch.zeros(128)
    mol_emb_list.append(emb)

mol_emb = torch.stack(mol_emb_list, dim=0)
print("Molecular (pseudo-3D) embedding shape:", mol_emb.shape)

# =========================
# 5. FINAL JOINT EMBEDDING + FAISS
# =========================

final_emb = torch.cat([text_emb, mol_emb], dim=1)
print("Final joint embedding shape:", final_emb.shape)

vecs = final_emb.detach().cpu().numpy().astype("float32")
d = vecs.shape[1]
index = faiss.IndexFlatIP(d)
faiss.normalize_L2(vecs)
index.add(vecs)

# =========================
# 6. RAG PIPELINE FUNCTIONS
# =========================

def embed_query(q: str) -> np.ndarray:
    q_emb = text_model.encode([q], convert_to_tensor=True)
    q_emb = q_emb.detach().cpu().numpy().astype("float32")
    pad = np.zeros((1, mol_emb.shape[1]), dtype="float32")
    full = np.concatenate([q_emb, pad], axis=1)
    faiss.normalize_L2(full)
    return full

def retrieve_top_k(question: str, k: int = 5):
    qv = embed_query(question)
    scores, idx = index.search(qv, k)
    return idx[0], scores[0]

def build_context(idx_list):
    rows = df_simple.iloc[idx_list]
    context_lines = []
    for _, r in rows.iterrows():
        context_lines.append(
            f"Product: {r[name_col]}\nComposition: {r['composition']}\n"
        )
    return "\n---\n".join(context_lines)

def simple_rag_answer(question: str, k: int = 5) -> str:
    idx_list, scores = retrieve_top_k(question, k=k)
    context = build_context(idx_list)
    answer = (
        f"Question: {question}\n\n"
        f"Relevant products and their raw materials:\n\n{context}\n\n"
        "These products illustrate which APIs and excipients occur together for this query."
    )
    return answer

# =========================
# 7. TEST RAG
# =========================

print(
    simple_rag_answer(
        "What raw materials are commonly present in paracetamol tablets?",
        k=5
    )
)



````

````python
import re

def extract_api_names(comp: str):
    """
    Very simple heuristic parser:
    - split by ';' and '+'
    - strip strengths in mg/ml/parcent
    - return cleaned ingredient names as 'APIs'
    """
    if not isinstance(comp, str):
        return []
    parts = re.split(r"[+;]", comp)
    apis = []
    for p in parts:
        p = p.strip()
        if not p:
            continue
        # remove strength info in brackets, e.g. (500mg/5ml)
        p_clean = re.sub(r"\([^)]*\)", "", p)
        p_clean = p_clean.strip()
        # skip empty or obviously non-API tokens
        if not p_clean:
            continue
        apis.append(p_clean)
    return apis

# quick sanity check
print(extract_api_names("Diclofenac (50mg) ;  Paracetamol (500mg) "))

# =========================
# 9. RAG "API" ENTRY FUNCTION
# =========================

def rag_api_call(question: str, k: int = 5):
    """
    API-style wrapper around the RAG pipeline.
    Returns a dict with:
      - question
      - retrieved products (name, composition, apis)
      - summary_context (a text summary of retrieved products)
    """
    idx_list, scores = retrieve_top_k(question, k=k)
    rows = df_simple.iloc[idx_list]

    results = []
    for i, (_, r) in enumerate(rows.iterrows()):
        apis = extract_api_names(r["composition"])
        results.append({
            "rank": i + 1,
            "score": float(scores[i]),
            "product_name": r[name_col],
            "composition": r["composition"],
            "apis": apis
        })

    # Generate a summary context similar to simple_rag_answer
    summary_context_text = build_context(idx_list)
    summary_context_text += "\n\nThese products illustrate which APIs and excipients occur together for this query."

    response = {
        "question": question,
        "top_k": k,
        "results": results,
        "summary_context": summary_context_text
    }
    return response

# =========================
# 10. DEMO "API" CALL
# =========================

resp = rag_api_call(
    "What raw materials are commonly present in paracetamol tablets?",
    k=5
)

from pprint import pprint
pprint(resp)
````

### Function to fetch SMILES from PubChem

````python
import requests
import time
import re # Ensure re is imported for regex operations

def get_smiles_from_pubchem(compound_name: str) -> str | None:
    """
    Fetches the canonical SMILES string for a given compound name from PubChem.
    Returns None if the compound is not found or an error occurs.
    """
    # Clean the name for better search results
    cleaned_name = compound_name.replace('(na)', '').replace(';', '').strip()
    # Further clean to handle cases like 'paracetamol (na) ;' -> 'paracetamol'
    cleaned_name = re.sub(r'\s+', ' ', cleaned_name).strip() # Replace multiple spaces with single, then strip

    if not cleaned_name:
        return None

    # Debug print: Show the cleaned name being queried
    print(f"DEBUG: Cleaned name for '{compound_name}': '{cleaned_name}'")

    # PubChem PUG/REST API endpoint for CID lookup by name
    cid_url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/name/{cleaned_name}/cids/json"
    data = None # Initialize data for error reporting
    smiles_data = None # Initialize smiles_data for error reporting

    try:
        response = requests.get(cid_url, timeout=10)
        response.raise_for_status() # Raise an exception for HTTP errors
        data = response.json()

        # Debug print: Show the data received from CID lookup
        print(f"DEBUG: CID lookup data for '{cleaned_name}': {data}")

        # Safely get CID
        cids = data.get("IdentifierList", {}).get("CID")
        if cids:
            cid = cids[0]

            # PubChem PUG/REST API endpoint for SMILES lookup by CID
            smiles_url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/{cid}/property/CanonicalSMILES/json"
            smiles_response = requests.get(smiles_url, timeout=10)
            smiles_response.raise_for_status() # Corrected: raise_for_status
            smiles_data = smiles_response.json()

            # Debug print: Show the raw smiles_data
            print(f"DEBUG: SMILES data for CID {cid}: {smiles_data}")

            # Safely get CanonicalSMILES
            properties = smiles_data.get("PropertyTable", {}).get("Properties")
            print(f"DEBUG: Extracted properties: {properties}") # Debug print
            if properties and len(properties) > 0:
                # Corrected: Use 'ConnectivitySMILES' instead of 'CanonicalSMILES'
                smiles = properties[0].get("ConnectivitySMILES")
                print(f"DEBUG: Extracted smiles: {smiles}") # Debug print
                if smiles:
                    return smiles
            else:
                print(f"WARNING: No SMILES properties found for CID {cid}. Data: {smiles_data}")

    except requests.exceptions.RequestException as e:
        print(f"Error fetching SMILES for {compound_name}: {e}")
    except Exception as e:
        # Catch any other unexpected errors, and print the last known data for debugging
        print(f"An unexpected error occurred for {compound_name}: {e}")
        if data:
            print(f"  Last CID data: {data}")
        if smiles_data:
            print(f"  Last SMILES data: {smiles_data}")

    return None

# Test with some examples from unique_ingredients
print("Fetching SMILES for 'paracetamol (na);':", get_smiles_from_pubchem("paracetamol (na);"))
print("Fetching SMILES for 'diclofenac (na);':", get_smiles_from_pubchem("diclofenac (na);"))
print("Fetching SMILES for 'amoxycillin (na);':", get_smiles_from_pubchem("amoxycillin (na);"))
print("Fetching SMILES for a non-existent compound 'nonexistentcompound123':", get_smiles_from_pubchem("nonexistentcompound123"))
````

## Generate RDKit Molecular Fingerprints

### Subtask:
For each obtained SMILES string, use RDKit to generate a molecular object and then compute a Morgan fingerprint. Store these fingerprints, possibly in a dictionary mapping ingredient names to their fingerprints.

````python
from rdkit import Chem
from rdkit.Chem import AllChem
from rdkit.Chem import rdFingerprintGenerator

def smiles_to_morgan_fingerprint(smiles_string, radius=2, nbits=2048):
    """
    Converts a SMILES string to a Morgan fingerprint.
    Returns a numpy array of integers representing the fingerprint, or None if conversion fails.
    """
    mol = Chem.MolFromSmiles(smiles_string)
    if mol is None:
        return None
    # Generate Morgan fingerprint using the recommended FingerprintGenerator
    # ECFP4 equivalent is Morgan fingerprint with radius=2
    fgen = rdFingerprintGenerator.GetMorganGenerator(radius=radius, fpSize=nbits)
    fp = fgen.GetFingerprint(mol)
    # Convert BitVector to numpy array
    return np.array(list(fp.ToBitString())).astype(int)

# Dictionary to store Morgan fingerprints for ingredients with successful SMILES retrieval
ingredient_to_morgan_fp = {}
failed_fp_generation = []

print("Generating Morgan fingerprints for retrieved SMILES...")
for ingredient, smiles in ingredient_to_smiles.items():
    fp = smiles_to_morgan_fingerprint(smiles)
    if fp is not None:
        ingredient_to_morgan_fp[ingredient] = fp
    else:
        failed_fp_generation.append(ingredient)

print(f"\nSuccessfully generated Morgan fingerprints for {len(ingredient_to_morgan_fp)} out of {len(ingredient_to_smiles)} ingredients with SMILES.")
print(f"Failed to generate fingerprints for {len(failed_fp_generation)} ingredients (invalid SMILES or other RDKit error).")

print("\nExample Morgan fingerprint (first ingredient with FP):")
if ingredient_to_morgan_fp:
    first_ing = next(iter(ingredient_to_morgan_fp.keys()))
    print(f"'{first_ing}': {ingredient_to_morgan_fp[first_ing][:10]}...") # Print first 10 bits
else:
    print("No fingerprints generated.")
````

````python
import requests
import time
import re # Ensure re is imported for regex operations

def get_smiles_from_pubchem(compound_name: str, max_retries: int = 5, initial_delay: float = 1.0) -> str | None:
    """
    Fetches the canonical SMILES string for a given compound name from PubChem,
    with retry logic and exponential backoff for transient API errors.
    Returns None if the compound is not found or an error occurs after retries.
    """
    # Clean the name for better search results
    cleaned_name = compound_name.replace('(na)', '').replace(';', '').strip()
    # Further clean to handle cases like 'paracetamol (na) ;' -> 'paracetamol'
    cleaned_name = re.sub(r'\s+', ' ', cleaned_name).strip()

    if not cleaned_name:
        return None

    data = None  # Initialize data for error reporting
    smiles_data = None  # Initialize smiles_data for error reporting

    for attempt in range(max_retries):
        delay = initial_delay * (2 ** attempt) # Exponential backoff

        # Debug print: Show the cleaned name being queried and current attempt
        print(f"DEBUG: Attempt {attempt + 1}/{max_retries} for '{compound_name}' (cleaned: '{cleaned_name}')")

        try:
            # PubChem PUG/REST API endpoint for CID lookup by name
            cid_url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/name/{cleaned_name}/cids/json"
            response = requests.get(cid_url, timeout=10)
            response.raise_for_status()  # Raise an exception for HTTP errors
            data = response.json()

            # Debug print: Show the data received from CID lookup
            print(f"DEBUG: CID lookup data for '{cleaned_name}': {data}")

            # Safely get CID
            cids = data.get("IdentifierList", {}).get("CID")
            if cids:
                cid = cids[0]

                # PubChem PUG/REST API endpoint for SMILES lookup by CID
                smiles_url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/{cid}/property/ConnectivitySMILES/json"
                smiles_response = requests.get(smiles_url, timeout=10)
                smiles_response.raise_for_status()  # Corrected: raise_for_status
                smiles_data = smiles_response.json()

                # Debug print: Show the raw smiles_data
                print(f"DEBUG: SMILES data for CID {cid}: {smiles_data}")

                # Safely get ConnectivitySMILES
                properties = smiles_data.get("PropertyTable", {}).get("Properties")
                print(f"DEBUG: Extracted properties: {properties}")  # Debug print
                if properties and len(properties) > 0:
                    smiles = properties[0].get("ConnectivitySMILES")
                    print(f"DEBUG: Extracted smiles: {smiles}")  # Debug print
                    if smiles:
                        return smiles # Successfully retrieved SMILES, exit function
                else:
                    print(f"WARNING: No SMILES properties found for CID {cid}. Data: {smiles_data}")
            else:
                print(f"INFO: No CID found for '{cleaned_name}'. Data: {data}")

        except requests.exceptions.RequestException as e:
            print(f"Error fetching SMILES for '{compound_name}' on attempt {attempt + 1}: {e}")
            if response.status_code in [503, 429]: # Transient errors
                print(f"Retrying in {delay:.1f} seconds...")
                time.sleep(delay)
                continue # Try again
            else:
                print(f"Non-retryable error, failing for '{compound_name}'.")
                return None # Fail immediately for other HTTP errors
        except Exception as e:
            # Catch any other unexpected errors, and print the last known data for debugging
            print(f"An unexpected error occurred for '{compound_name}' on attempt {attempt + 1}: {e}")
            if data:
                print(f"  Last CID data: {data}")
            if smiles_data:
                print(f"  Last SMILES data: {smiles_data}")
            return None # Fail for unexpected errors

    print(f"ERROR: Failed to retrieve SMILES for '{compound_name}' after {max_retries} attempts.")
    return None # Return None if all retries fail

# Test with some examples (including one that might fail for retries if hammered too much)
print("Fetching SMILES for 'paracetamol':", get_smiles_from_pubchem("paracetamol"))
print("Fetching SMILES for 'diclofenac':", get_smiles_from_pubchem("diclofenac"))
print("Fetching SMILES for a non-existent compound 'nonexistentcompound123':", get_smiles_from_pubchem("nonexistentcompound123"))
print("Fetching SMILES for 'glucose' with retries:", get_smiles_from_pubchem("glucose", max_retries=3, initial_delay=0.5))
````
